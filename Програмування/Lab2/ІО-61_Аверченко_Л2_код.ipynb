### Завдання 1

```python
import math

x = float(input("Введіть змінну x: "))
y = float(input("Введіть змінну y: "))

if (3.8 * x + y) > 0 and (3.8 * x + y) != 1:
    R = ((math.e ** (2 * x) + math.sin(y))
        / (math.log(3.8 * x + y, math.e)))
else:
    R = "Вираз неможливо обчислити для заданих значень"

print(R)
```

### Завдання 2

```python
a = float(input("Введіть довільне а: "))
b = float(input("Введіть довільне b: "))

sumOfSquared = a ** 2 + b ** 2
squareOfSum = (a + b) ** 2

print("Сума квадратів a, b: ", sumOfSquared)
print("Квадрат суми a, b: ", squareOfSum)

if sumOfSquared > squareOfSum:
    print("Сума квадратів більша за квадрат суми")
elif sumOfSquared < squareOfSum:
    print("Квадрат суми більше за суму квадратів")
else:
    print("Квадрат суми дорівнює сумі квадратів")
```

### Завдання 3

```python
a = int(input("Введіть ціле число a: "))
b = int(input("Введіть ціле число b (b > a): "))

if b <= a:
    print("b повинно бути більше a")
else:
    numbersBetween = range(a, b + 1)
    N = len(numbersBetween)

    print("Числа між a та b:", end=" ")

    for i in numbersBetween:
        print(i, end=", ")

    print()
    print("Кількість чисел", N)
```
