### Завдання 1

```python
from functools import reduce
import random

arr1 = [random.randint(0, 100) for _ in range(10)]
arr2 = [random.randint(0, 100) for _ in range(10)]
result = []
print(f"Список 1: {arr1}\nСписок 2: {arr2}")

for i in range(0, min(len(arr1), len(arr2)), 2):
  result.append(arr1[i + 1])
  result.append(arr2[i])

print(f"Новий список: {result}")
result = random.sample(result, len(result) // 2)

print(f"Результуючий рядок: {', '.join(map(str, result))}")
```

### Завдання 2

```python
import random

m = int(input("Введіть m: "))
n = int(input("Введіть n: "))

matrixA = []
resultMatrix = []

for i in range(m):
  row = [random.randint(0, 10) for _ in range(n)]
  matrixA.append(row)

print(f"Матриця А:\n{'\n'.join(map(str, matrixA))}")

for row in matrixA:
  if (len(set(row)) == len(row)):
    resultMatrix.append(row)

matrixA = resultMatrix
print(f"Матриця А після перетворення:\n{'\n'.join(map(str, matrixA))}")
```
